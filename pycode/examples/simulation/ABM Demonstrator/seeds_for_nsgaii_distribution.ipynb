```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("input/pop8_sel_particles.csv")

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

parameters = ["seed","initE", "damp_time", "damp_lvl"]

for ax, param in zip(axes, parameters):
    ax.hist(df[param], bins=15, edgecolor="black")
    ax.set_title(param)
    ax.set_xlabel(param)
    ax.set_ylabel("Frequency")
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
